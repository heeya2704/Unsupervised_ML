## Task 1: Dataset Loading & First 10 Rows
Load the Customer Segmentation Tutorial Dataset from Kaggle (`mall_customers.csv`) and display the first 10 rows using pandas.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans

# Load dataset
file_path = '../Datasets/mall_customers.csv' if os.path.exists('../Datasets/mall_customers.csv') else 'Datasets/mall_customers.csv'
df = pd.read_csv(file_path)

print(f"Dataset Shape: {df.shape[0]} rows x {df.shape[1]} columns")
df.head(10)
```

--- 
## Task 2: Missing Values Audit
Check for missing values in `Age`, `Annual Income (k$)`, and `Spending Score (1-100)` columns and print total missing counts for each.

```python
target_cols = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
missing_counts = df[target_cols].isnull().sum()

print("Missing Values Summary:")
for col in target_cols:
    print(f"  • {col:<30}: {missing_counts[col]} missing values")
    
print(f"\nTotal Missing Values across target columns: {missing_counts.sum()}")
```

--- 
## Task 3: Scatterplot of Annual Income vs Spending Score
Create a scatterplot of `Annual Income (k$)` vs `Spending Score (1-100)` using Seaborn / Matplotlib.

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Gender',
    palette={'Male': '#1f77b4', 'Female': '#e377c2'},
    s=80,
    alpha=0.85,
    edgecolor='k'
)

plt.title("Mall Customer Distribution: Annual Income vs Spending Score", fontsize=13, fontweight='bold')
plt.xlabel("Annual Income (k$)", fontsize=11, fontweight='bold')
plt.ylabel("Spending Score (1-100)", fontsize=11, fontweight='bold')
plt.tight_layout()
plt.savefig('task_3_customer_scatterplot.png', dpi=300)
plt.show()
```

--- 
## Task 4: Correlation Matrix Heatmap & Interpretation
Generate a correlation heatmap for numerical features and interpret which two features are most strongly correlated.

```python
numeric_df = df.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, fmt=".3f", cmap="coolwarm", vmin=-1, vmax=1, linewidths=1)
plt.title("Correlation Matrix Heatmap (Numerical Mall Features)", fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('task_4_correlation_heatmap.png', dpi=300)
plt.show()

print("Strongest Correlated Feature Pair Interpretation:")
print("Excluding the CustomerID identifier, 'Age' and 'Spending Score (1-100)' exhibit the strongest linear correlation (r = -0.327).")
print("Interpretation: This negative correlation indicates that younger customers generally have higher spending scores, whereas older customers tend to have lower spending scores.")
```

--- 
## Task 5: K-Means Customer Segmentation (k=3, random_state=42)
Apply K-Means clustering ($k=3$, `random_state=42`) using `Annual Income (k$)` and `Spending Score (1-100)`, assign the `Cluster` column, and plot customer segments.

```python
X = df[['Annual Income (k$)', 'Spending Score (1-100)']].values

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X)
centroids = kmeans.cluster_centers_

print("Cluster Counts:")
print(df['Cluster'].value_counts().sort_index())

print("\nLearned Centroids:")
for idx, center in enumerate(centroids):
    print(f"Cluster {idx}: Income = {center[0]:.2f} k$ | Spending Score = {center[1]:.2f}")

plt.figure(figsize=(9, 6))
palette = sns.color_palette("Set1", n_colors=3)
for c_id in range(3):
    c_df = df[df['Cluster'] == c_id]
    plt.scatter(c_df['Annual Income (k$)'], c_df['Spending Score (1-100)'], color=palette[c_id], label=f'Cluster {c_id}', alpha=0.8, s=70)

plt.scatter(centroids[:, 0], centroids[:, 1], color='red', marker='X', s=250, edgecolor='black', linewidth=2, label='Centroids', zorder=10)
plt.title("K-Means Customer Segmentation (k=3, random_state=42)", fontsize=13, fontweight='bold')
plt.xlabel("Annual Income (k$)", fontsize=11, fontweight='bold')
plt.ylabel("Spending Score (1-100)", fontsize=11, fontweight='bold')
plt.legend(loc='upper right')
plt.tight_layout()
plt.savefig('task_5_kmeans_customer_clusters.png', dpi=300)
plt.show()

df.head(10)
```
